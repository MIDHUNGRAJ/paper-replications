```python
import torch
import torch.nn as nn 
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
from torch.utils.data import DataLoader
import matplotlib.pyplot as plt
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
```

```python
def conv1x1(input, output, stride=1):
    return nn.Conv2d(
        input, output, kernel_size=1, stride=stride
    ) 


def conv3x3(input, output, stride=1):
    return nn.Conv2d(
        input, output, kernel_size=3, stride=stride
    )
```

```python
class bottleneck(nn.Module):
    expansion = 4
    def __init__(self, input, output, stride, shortcut=None):
        self.conv1 = conv1x1(input=input, output=output)
        self.bn1 = nn.BatchNorm2d(output)

        self.conv2 = conv3x3(input=output, output=output, stride=stride)
        self.bn2 = nn.BatchNorm2d(output)

        self.conv3 = conv1x1(input=output, output=output*expansion)
        self.bn3 = nn.BatchNorm2d(output*expansion)

        self.shortcut = shortcut
        self.relu = nn.ReLU()

    def forward(self, x):
        identity = x

        out = self.conv1(x)
        out = self.bn1(out)
        out = self.relu(out)

        out = self.conv2(out)
        out = self.bn2(out)
        out = self.relu(out)

        out = self.conv3(out)
        out = self.bn3(out)
        out = self.relu(out)

        if self.shortcut is not None:
            identity = self.shortcut(x)

        out += identity
        out = self.relu(out)
        return out

```
